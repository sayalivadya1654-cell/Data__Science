# Getting Started With Machine Learning

# Linear Regression  
- Linear Regression is a supervised machine learning algorithm used to predict a continuous numerical value.
- It learns a linear relationship between input features and the target using coefficients and an intercept.

- | Input (X)           | Output (Y)  |
| ------------------- | ----------- |
| Years of Experience | Salary      |
| Area of House       | House Price |
| Advertising Cost    | Sales       |
| Hours Studied       | Marks       |
- Experience = 5 years → Salary = ₹60,000
- The model learns the relationship between X and Y and predicts Y for new X.

## Why is it called "Linear"?
- Because it tries to find the best straight line through the data.

## Linear Regression Equation ⭐
- Simple Linear Regression
$$ \boxed{y = mx+c} $$

Where:

Symbol	Meaning
y	predicted output
x	input
m	slope
c	intercept

In machine learning, you will also see:

$$ \boxed{\hat y = b_0+b_1x} $$

Where:

- \(\hat y) = predicted value
- \(b_0\) = intercept
- \(b_1\) = coefficient/slope
- \(x\) = input

## What is Slope?
- Slope tells us how much Y changes when X increases by 1 unit.
- formula: $$ m = Δx/Δy $$
Example:
- Experience = 1 year → Salary = 500
- Experience = 2 years → Salary = 700

Then:

$$ m = \frac{700-500}{2-1} $$ $$ m=200 $$

- So salary increases by 200 units for every 1-year increase in experience.
	​


## What is Intercept?
- Intercept is the predicted value of Y when:

$$ x=0 $$

For:

$$ y=200x+300 $$

When x = 0:

$$ y=300 $$

- So 300 is the intercept.
