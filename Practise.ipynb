```python
# Program to find LCM
def compute_lcm(x, y):
    if x > y:
        greater = x
    else:
        greater = y

    while True:
        if (greater % x == 0) and (greater % y == 0):
            return greater
        greater += 1

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
print("The LCM is", compute_lcm(num1, num2))
```

```python
# Program to find HCF of two numbers
def calculate_hcf(x, y):
    if x > y:
        smaller = y
    else:
        smaller = x

    for i in range(1, smaller + 1):
        if (x % i == 0) and (y % i == 0):
            hcf = i
    return hcf
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
print("The HCF is", calculate_hcf(num1, num2))
```

```python
# find common elements
a = [1,2,3,4,5]
b = [1,3,5,6,7]
print(list(set(a) & set(b)))
set(a) & set(b)     # common elements
set(a) | set(b)     # all unique elements
set(a) - set(b)     # elements only in a
set(b) - set(a)     # elements only in b
```
