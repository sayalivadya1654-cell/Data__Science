```python
# Program to find LCM
def compute_lcm(x, y):
    if x > y:
        greater = x
    else:
        greater = y

    while True:
        if (greater % x == 0) and (greater % y == 0):
            return greater
        greater += 1

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
print("The LCM is", compute_lcm(num1, num2))
```

```python
# Program to find HCF of two numbers
def calculate_hcf(x, y):
    if x > y:
        smaller = y
    else:
        smaller = x

    for i in range(1, smaller + 1):
        if (x % i == 0) and (y % i == 0):
            hcf = i
    return hcf
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
print("The HCF is", calculate_hcf(num1, num2))
```

# indexing and slicing

```python
# indexing in 1d
import numpy as np
arr1 = np.array([[1,2],[3,4]])
print(arr1[0])
print(arr1[1])
arr2 = np.array([10,20,30,40,50])
print(arr2[-1])
print(arr2[0])
print(arr2[2])
```

```python
# indexing in 2d
arr1 = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr1[0,0])
print(arr1[0,1])
print(arr1[0,2])
print(arr1[1,0])
print(arr1[1,1])
print(arr1[1,2])
```

```python
# slicing in 1d
arr1 = np.array([1,2,3,4,5,6])
print(arr1[ : ]) # entire row
print(arr1[:3]) # first 3 elements
print(arr1[2:]) # from index 2
print(arr1[1:5]) # middle values
print(arr1[::2]) # every second
print(arr1[::-1]) # reverse array
```

```python
# slicing in 2d
arr1 = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
print(arr1[0,:]) # entire first row
print(arr1[1,:]) # entire second row
print(arr1[2,:]) # entire third row
print(arr1[0:2,:]) # first 2 rows
print(arr1[:,1:]) # only 2nd and first column
print(arr1[1:,0:]) # last two rows
```

```python
# reshape()
arr = np.array([10,20,30,40,50,60])
new_arr = arr.reshape(2,3)
print(new_arr)
```

```python
arr = np.arange(6)
new_arr = arr.reshape(2,3)
print(new_arr)
```

```python
arr = np.arange(10)
new_arr = arr.reshape(2,5)
print(new_arr)
arr1 = np.arange(16)
new_arr = arr1.reshape(4,4)
print(new_arr)
arr = np.arange(9)
new_arr = arr.reshape(3,3)
print(new_arr)
```

```python
# flatten
arr = np.array([
    [1,2,3],
    [4,5,6]
])
new_arr = arr.flatten()
print(new_arr)
```

```python
# ravel()
arr = np.array([
    [1,2,3],
    [4,5,6]
])
new_arr = arr.ravel()
print(new_arr)
```

```python
# transpose
arr = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr.T)
arr1 = np.array([
    [1,2],
    [3,4]
])
print(arr.T)
```

```python
# mathematical operation
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a + b) # element wise addition
print(a * b) # element wise multiplication
print(a @ b) # matrix multiplication
print(np.sum(a)) # sum
print(np.max(a)) # max number
print(np.min(b)) # min from b
print(np.mean(a)) # mean
print(np.median(a)) # median
print(np.std(a)) # std deviation
print(np.mean(b))
```

```python
# concatenate
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
np.concatenate((a,b))
```

# matplotlib

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.plot(x,y, label = "sales")
plt.title("basic graph")
plt.xlabel("candidates")
plt.ylabel("marks")
plt.xticks(rotation = 45)
plt.yticks(rotation = 30)
plt.grid(axis="x")
plt.grid(axis="y")
# plt.grid()       # used to print both x and y axis grid
plt.legend()
plt.figure(figsize=(5,8))
plt.tight_layout()
plt.savefig("sales.png")
plt.show()
```
