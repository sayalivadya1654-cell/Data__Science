# Getting started with **SQL**

### SQL: Structured Query Language
-SQL (Structured Query Language) is a standard language used to communicate with relational databases. It helps store, retrieve, update, and manage data.

Database
-A database is an organized collection of related data.

Types of Databases
-Databases can be classified based on how they store and manage data.

1. Relational Database (RDBMS)
-Stores data in tables (rows and columns) and uses SQL.

-Features
1. Structured data
2. Relationships using Primary Keys and Foreign Keys
3. ACID-compliant transactions

-Examples
MySQL
PostgreSQL
Oracle Database
Microsoft SQL Server

-Use Case: Banking systems, ERP, student management systems.

2. NoSQL Database
-Stores data in formats other than tables.

-Features
1. Flexible schema
2. High scalability
3. Handles large volumes of data

-Examples
MongoDB
Apache Cassandra
Redis

-Use Case: Social media, real-time analytics, big data applications.

3. Hierarchical Database
-Stores data in a tree-like structure with parent-child relationships.

-Structure
```
Company
 ├─ Department
 │   ├─ Employee
 │   └─ Employee
```

-Features
1. Fast access for hierarchical data
2. One-to-many relationships

-Use Case: File systems, organization charts.

4. Network Database
-Stores data as a network where a record can have multiple parent and child records.

-Features
1. Many-to-many relationships
2. More flexible than hierarchical databases

-Use Case: Telecommunications and logistics systems.

5. Object-Oriented Database (OODBMS)
-Stores data as objects similar to object-oriented programming languages.

-Features
1. Supports classes and inheritance
2. Suitable for complex applications

-Examples
ObjectDB

-Use Case: CAD, engineering, multimedia applications.

6. Document Database
-A type of NoSQL database that stores data as documents (usually JSON).

-Example Document
```
{
  "id": 101,
  "name": "Ravi",
  "city": "Mumbai"
}
```
-Examples
MongoDB
CouchDB

-Use Case: Content management systems and web applications.

7. Key-Value Database
-Stores data as key-value pairs.

-Example
```
Name → Ravi
Age  → 20
```

-Examples
Redis
Amazon DynamoDB

-Use Case: Caching, session storage.

8. Column-Oriented Database
-Stores data by columns instead of rows.

-Examples
Apache HBase
Google Bigtable

-Use Case: Data warehousing and analytics.

9. Graph Database
-Stores data as nodes and relationships (edges).

-Features
1. Excellent for connected data
2. Fast relationship traversal

-Examples
Neo4j
Amazon Neptune

-Use Case: Social networks, recommendation engines, fraud detection.

Categories of SQL Commands

1. DDL (Data Definition Language)
   - Used to define or modify database structures.
   - Commands: CREATE, ALTER, DROP, TRUNCATE, RENAME

| Command           | Purpose                       | Generic Syntax                                                 |
| ----------------- | ----------------------------- | -------------------------------------------------------------- |
| `CREATE DATABASE` | Creates a database            | `CREATE DATABASE database_name;`                               |
| `CREATE TABLE`    | Creates a table               | `CREATE TABLE table_name (column datatype);`                   |
| `ALTER TABLE`     | Modifies a table structure    | `ALTER TABLE table_name ADD/MODIFY/DROP column_name datatype;` |
| `DROP DATABASE`   | Deletes a database            | `DROP DATABASE database_name;`                                 |
| `DROP TABLE`      | Deletes a table               | `DROP TABLE table_name;`                                       |
| `TRUNCATE TABLE`  | Removes all rows from a table | `TRUNCATE TABLE table_name;`                                   |
| `RENAME TABLE`    | Renames a table               | `ALTER TABLE old_name RENAME TO new_name;`                     |

2. DML (Data Manipulation Language)
   - Used to manipulate data in tables.
   - Commands: INSERT, UPDATE, DELETE

| Command  | Purpose                                    | Generic Syntax                                                              |
| -------- | ------------------------------------------ | --------------------------------------------------------------------------- |
| `INSERT` | Inserts new records                        | `INSERT INTO table_name (col1,col2) VALUES (val1,val2);`                    |
| `UPDATE` | Updates existing records                   | `UPDATE table_name SET column_name = value WHERE condition;`                |
| `DELETE` | Deletes records                            | `DELETE FROM table_name WHERE condition;`                                   |
| `MERGE`  | Inserts, updates, or deletes conditionally | `MERGE INTO target USING source ON condition WHEN MATCHED THEN UPDATE ...;` |

3. DCL (Data Control Language)
   - Used to control permissions.
   - Commands: GRANT, REVOKE

| Command  | Purpose             | Generic Syntax                          |
| -------- | ------------------- | --------------------------------------- |
| `GRANT`  | Gives permissions   | `GRANT privilege ON object TO user;`    |
| `REVOKE` | Removes permissions | `REVOKE privilege ON object FROM user;` |

4. DQL (Data Query Language)
   - Used to retrieve data.
   - Command: SELECT

| Command    | Purpose            | Generic Syntax                                 |
| ---------- | ------------------ | ---------------------------------------------- |
| `SELECT`   | Retrieves data     | `SELECT column_name FROM table_name;`          |
| `DISTINCT` | Removes duplicates | `SELECT DISTINCT column_name FROM table_name;` |

5. TCL (Transaction Control Language)
   - Used to manage transactions.
   - Commands: COMMIT, ROLLBACK, SAVEPOINT

| Command             | Purpose                       | Generic Syntax                      |
| ------------------- | ----------------------------- | ----------------------------------- |
| `COMMIT`            | Saves transaction permanently | `COMMIT;`                           |
| `ROLLBACK`          | Undoes transaction            | `ROLLBACK;`                         |
| `SAVEPOINT`         | Creates a checkpoint          | `SAVEPOINT savepoint_name;`         |
| `RELEASE SAVEPOINT` | Removes a savepoint           | `RELEASE SAVEPOINT savepoint_name;` |

Constraints
-Constraints are rules applied to table columns to enforce data integrity and ensure that only valid data is stored in the database.

| Constraint      | Description                                  | Short Syntax                                        |
| --------------- | -------------------------------------------- | --------------------------------------------------- |
| **NOT NULL**    | Prevents NULL values in a column.            | `Name VARCHAR(50) NOT NULL`                         |
| **UNIQUE**      | Ensures all values in a column are unique.   | `Email VARCHAR(100) UNIQUE`                         |
| **PRIMARY KEY** | Uniquely identifies each row in a table.     | `EmpID INT PRIMARY KEY`                             |
| **FOREIGN KEY** | Creates a relationship between two tables.   | `FOREIGN KEY(DeptID) REFERENCES Department(DeptID)` |
| **CHECK**       | Restricts values based on a condition.       | `CHECK (Age >= 18)`                                 |
| **DEFAULT**     | Assigns a default value if none is provided. | `Status VARCHAR(20) DEFAULT 'Active'`               |

Aggregate Functions

| ---------- | -------------------- |
| `COUNT()`  | Counts rows          |
| `SUM()`    | Calculates total     |
| `AVG()`    | Calculates average   |
| `MIN()`    | Finds smallest value |

| `GROUP BY` | Groups rows          |

SQL Operators
-Operators are special symbols or keywords used to perform operations on data and specify conditions in SQL queries.

categories of operators

| Category                 | Description                                                         |
| ------------------------ | ------------------------------------------------------------------- |
| **Arithmetic Operators** | Perform mathematical calculations on numeric values.                |
| **Comparison Operators** | Compare two values and return TRUE or FALSE.                        |
| **Logical Operators**    | Combine or modify multiple conditions.                              |
| **Special Operators**    | Perform specific filtering, pattern matching, and existence checks. |
| **Set Operators**        | Combine results from two or more SELECT queries.                    |

1. Arithmetic Operators
| Operator | Description                      | Short Syntax     |
| -------- | -------------------------------- | ---------------- |
| `+`      | Adds two values                  | `SELECT 10 + 5;` |
| `-`      | Subtracts one value from another | `SELECT 10 - 5;` |
| `*`      | Multiplies values                | `SELECT 10 * 5;` |
| `/`      | Divides values                   | `SELECT 10 / 5;` |
| `%`      | Returns remainder after division | `SELECT 10 % 3;` |

2. Comparison Operators
| Operator | Description                 | Short Syntax            |
| -------- | --------------------------- | ----------------------- |
| `=`      | Equal to                    | `WHERE Salary = 50000`  |
| `!=`     | Not equal to                | `WHERE Salary != 50000` |
| `<>`     | Not equal to (standard SQL) | `WHERE Salary <> 50000` |
| `>`      | Greater than                | `WHERE Salary > 50000`  |
| `<`      | Less than                   | `WHERE Salary < 50000`  |
| `>=`     | Greater than or equal to    | `WHERE Salary >= 50000` |
| `<=`     | Less than or equal to       | `WHERE Salary <= 50000` |

3. Logical Operators
| Operator | Description                                    | Short Syntax                       |
| -------- | ---------------------------------------------- | ---------------------------------- |
| `AND`    | Returns TRUE if all conditions are true        | `WHERE Dept='IT' AND Salary>50000` |
| `OR`     | Returns TRUE if at least one condition is true | `WHERE Dept='IT' OR Dept='HR'`     |
| `NOT`    | Reverses the result of a condition             | `WHERE NOT Dept='IT'`              |

4. Special Operators
| Operator  | Description                                 | Short Syntax                           |
| --------- | ------------------------------------------- | -------------------------------------- |
| `IN`      | Checks whether a value exists in a list     | `WHERE Dept IN ('IT','HR')`            |
| `BETWEEN` | Checks whether a value falls within a range | `WHERE Salary BETWEEN 30000 AND 60000` |
| `LIKE`    | Searches for a pattern in text              | `WHERE Name LIKE 'A%'`                 |
| `IS NULL` | Checks for NULL values                      | `WHERE Phone IS NULL`                  |
| `EXISTS`  | Checks if a subquery returns rows           | `WHERE EXISTS (SELECT 1 FROM Orders)`  |

5. Set Operators
| Operator    | Description                                 | Short Syntax                                      |
| ----------- | ------------------------------------------- | ------------------------------------------------- |
| `UNION`     | Combines results and removes duplicates     | `SELECT City FROM A UNION SELECT City FROM B`     |
| `UNION ALL` | Combines results including duplicates       | `SELECT City FROM A UNION ALL SELECT City FROM B` |
| `INTERSECT` | Returns common rows from both queries       | `SELECT City FROM A INTERSECT SELECT City FROM B` |
| `EXCEPT`    | Returns rows from first query not in second | `SELECT City FROM A EXCEPT SELECT City FROM B`    |


Joins in SQL
-Joins are used to combine data from two or more tables based on a related column (usually a Primary Key and Foreign Key).

Types of join

| Join Type      | Description                                                                        | Short Syntax                                                   |
| -------------- | ---------------------------------------------------------------------------------- | -------------------------------------------------------------- |
| **INNER JOIN** | Returns only matching records from both tables.                                    | `SELECT * FROM A INNER JOIN B ON A.id = B.id;`                 |
| **LEFT JOIN**  | Returns all records from the left table and matching records from the right table. | `SELECT * FROM A LEFT JOIN B ON A.id = B.id;`                  |
| **RIGHT JOIN** | Returns all records from the right table and matching records from the left table. | `SELECT * FROM A RIGHT JOIN B ON A.id = B.id;`                 |
| **FULL JOIN**  | Returns all records when there is a match in either table.                         | `SELECT * FROM A FULL JOIN B ON A.id = B.id;`                  |
| **CROSS JOIN** | Returns every possible combination of rows from both tables.                       | `SELECT * FROM A CROSS JOIN B;`                                |
| **SELF JOIN**  | Joins a table with itself.                                                         | `SELECT * FROM Emp E1 JOIN Emp E2 ON E1.ManagerID = E2.EmpID;` |


CASE Statement
-The CASE statement is used to add conditional logic in SQL. It works like IF-ELSE or switch-case in programming languages and returns a value based on specified conditions.

```
CASE
  WHEN condition THEN result
  ELSE default-result
END
```

| Keyword | Purpose                     |
| ------- | --------------------------- |
| `CASE`  | Starts conditional logic    |
| `WHEN`  | Specifies condition/value   |
| `THEN`  | Result if condition is true |
| `ELSE`  | Default result if no match  |
| `END`   | Ends the CASE statement     |

Case Types

| CASE Type         | Description                                   | Short Syntax                                            |
| ----------------- | --------------------------------------------- | ------------------------------------------------------- |
| **Simple CASE**   | Compares one expression with multiple values. | `CASE DeptID WHEN 101 THEN 'IT' WHEN 102 THEN 'HR' END` |
| **Searched CASE** | Evaluates multiple conditions using `WHEN`.   | `CASE WHEN Salary > 50000 THEN 'High' ELSE 'Low' END`   |

Window function
-Window Functions perform calculations across a set of rows related to the current row while retaining individual rows. Unlike GROUP BY, they do not collapse multiple rows into a single result.

1. Ranking Functions
-Assign rankings or sequence numbers to rows.

| Function         | Description                            | Short Syntax                              |
| ---------------- | -------------------------------------- | ----------------------------------------- |
| **ROW_NUMBER()** | Unique sequential number for each row. | `ROW_NUMBER() OVER(ORDER BY Salary DESC)` |
| **RANK()**       | Rank with gaps for ties.               | `RANK() OVER(ORDER BY Salary DESC)`       |
| **DENSE_RANK()** | Rank without gaps for ties.            | `DENSE_RANK() OVER(ORDER BY Salary DESC)` |
| **NTILE(n)**     | Divides rows into n equal groups.      | `NTILE(4) OVER(ORDER BY Salary)`          |

2. Analytical Functions
-Access values from preceding or following rows.

| Function   | Description                      | Short Syntax                        |
| ---------- | -------------------------------- | ----------------------------------- |
| **LAG()**  | Returns value from previous row. | `LAG(Salary) OVER(ORDER BY EmpID)`  |
| **LEAD()** | Returns value from next row.     | `LEAD(Salary) OVER(ORDER BY EmpID)` |

3. Aggregate Window Functions
-Perform aggregate calculations without grouping rows.

| Function           | Description               | Short Syntax                            |
| ------------------ | ------------------------- | --------------------------------------- |
| **SUM() OVER()**   | Running/Cumulative total. | `SUM(Salary) OVER(ORDER BY EmpID)`      |
| **AVG() OVER()**   | Running average.          | `AVG(Salary) OVER(ORDER BY EmpID)`      |
| **COUNT() OVER()** | Counts rows in window.    | `COUNT(*) OVER(PARTITION BY DeptID)`    |
| **MIN() OVER()**   | Minimum value in window.  | `MIN(Salary) OVER(PARTITION BY DeptID)` |
| **MAX() OVER()**   | Maximum value in window.  | `MAX(Salary) OVER(PARTITION BY DeptID)` |

4. Value Functions
-Return specific values from the window frame.

| Function          | Description                    | Short Syntax                                |
| ----------------- | ------------------------------ | ------------------------------------------- |
| **FIRST_VALUE()** | Returns first value in window. | `FIRST_VALUE(Salary) OVER(ORDER BY Salary)` |
| **LAST_VALUE()**  | Returns last value in window.  | `LAST_VALUE(Salary) OVER(ORDER BY Salary)`  |
| **NTH_VALUE()**   | Returns nth value in window.   | `NTH_VALUE(Salary,2) OVER(ORDER BY Salary)` |
