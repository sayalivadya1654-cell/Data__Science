# Getting started with **SQL**

# SQL Notes

## 1. What is SQL?
**SQL (Structured Query Language)** is a standard language used to communicate with relational databases. It helps store, retrieve, update, and manage data.

**Database** — An organized collection of related data.

---

## 2. Types of Databases

### a) Relational Database (RDBMS)
Stores data in tables (rows and columns) and uses SQL.

- **Features:** Structured data, relationships via Primary/Foreign Keys, ACID-compliant transactions
- **Examples:** MySQL, PostgreSQL, Oracle Database, Microsoft SQL Server
- **Use case:** Banking systems, ERP, student management systems

### b) NoSQL Database
Stores data in formats other than tables.

- **Features:** Flexible schema, high scalability, handles large data volumes
- **Examples:** MongoDB, Apache Cassandra, Redis
- **Use case:** Social media, real-time analytics, big data applications

### c) Hierarchical Database
Stores data in a tree-like structure with parent-child relationships.

```
Company
 ├─ Department
 │   ├─ Employee
 │   └─ Employee
```
- **Features:** Fast access for hierarchical data, one-to-many relationships
- **Use case:** File systems, organization charts

### d) Network Database
Stores data as a network where a record can have multiple parent and child records.

- **Features:** Many-to-many relationships, more flexible than hierarchical DBs
- **Use case:** Telecommunications and logistics systems

### e) Object-Oriented Database (OODBMS)
Stores data as objects, similar to OOP languages.

- **Features:** Supports classes and inheritance, suitable for complex applications
- **Example:** ObjectDB
- **Use case:** CAD, engineering, multimedia applications

### f) Document Database
A NoSQL database that stores data as documents (usually JSON).

```json
{
  "id": 101,
  "name": "Ravi",
  "city": "Mumbai"
}
```
- **Examples:** MongoDB, CouchDB
- **Use case:** Content management systems, web applications

### g) Key-Value Database
Stores data as key-value pairs.

```
Name → Ravi
Age  → 20
```
- **Examples:** Redis, Amazon DynamoDB
- **Use case:** Caching, session storage

### h) Column-Oriented Database
Stores data by columns instead of rows.

- **Examples:** Apache HBase, Google Bigtable
- **Use case:** Data warehousing and analytics

### i) Graph Database
Stores data as nodes and relationships (edges).

- **Features:** Excellent for connected data, fast relationship traversal
- **Examples:** Neo4j, Amazon Neptune
- **Use case:** Social networks, recommendation engines, fraud detection

## 3. Categories of SQL Commands

### DDL — Data Definition Language
*Defines or modifies database structures.*
Commands: `CREATE`, `ALTER`, `DROP`, `TRUNCATE`, `RENAME`

| Command | Purpose | Syntax |
|---|---|---|
| CREATE DATABASE | Creates a database | `CREATE DATABASE database_name;` |
| CREATE TABLE | Creates a table | `CREATE TABLE table_name (column datatype);` |
| ALTER TABLE | Modifies table structure | `ALTER TABLE table_name ADD/MODIFY/DROP column_name datatype;` |
| DROP DATABASE | Deletes a database | `DROP DATABASE database_name;` |
| DROP TABLE | Deletes a table | `DROP TABLE table_name;` |
| TRUNCATE TABLE | Removes all rows | `TRUNCATE TABLE table_name;` |
| RENAME TABLE | Renames a table | `ALTER TABLE old_name RENAME TO new_name;` |

### DML — Data Manipulation Language
*Manipulates data in tables.*
Commands: `INSERT`, `UPDATE`, `DELETE`

| Command | Purpose | Syntax |
|---|---|---|
| INSERT | Inserts new records | `INSERT INTO table_name (col1,col2) VALUES (val1,val2);` |
| UPDATE | Updates existing records | `UPDATE table_name SET column_name = value WHERE condition;` |
| DELETE | Deletes records | `DELETE FROM table_name WHERE condition;` |
| MERGE | Insert/Update/Delete conditionally | `MERGE INTO target USING source ON condition WHEN MATCHED THEN UPDATE ...;` |

### DCL — Data Control Language
*Controls permissions.*
Commands: `GRANT`, `REVOKE`

| Command | Purpose | Syntax |
|---|---|---|
| GRANT | Gives permissions | `GRANT privilege ON object TO user;` |
| REVOKE | Removes permissions | `REVOKE privilege ON object FROM user;` |

### DQL — Data Query Language
*Retrieves data.*
Command: `SELECT`

| Command | Purpose | Syntax |
|---|---|---|
| SELECT | Retrieves data | `SELECT column_name FROM table_name;` |
| DISTINCT | Removes duplicates | `SELECT DISTINCT column_name FROM table_name;` |

### TCL — Transaction Control Language
*Manages transactions.*
Commands: `COMMIT`, `ROLLBACK`, `SAVEPOINT`

| Command | Purpose | Syntax |
|---|---|---|
| COMMIT | Saves transaction permanently | `COMMIT;` |
| ROLLBACK | Undoes transaction | `ROLLBACK;` |
| SAVEPOINT | Creates a checkpoint | `SAVEPOINT savepoint_name;` |
| RELEASE SAVEPOINT | Removes a savepoint | `RELEASE SAVEPOINT savepoint_name;` |

## 5. Constraints
Rules applied to columns to enforce data integrity.

| Constraint | Description | Syntax |
|---|---|---|
| NOT NULL | Prevents NULL values | `Name VARCHAR(50) NOT NULL` |
| UNIQUE | Ensures unique values | `Email VARCHAR(100) UNIQUE` |
| PRIMARY KEY | Uniquely identifies each row | `EmpID INT PRIMARY KEY` |
| FOREIGN KEY | Links two tables | `FOREIGN KEY(DeptID) REFERENCES Department(DeptID)` |
| CHECK | Restricts values by condition | `CHECK (Age >= 18)` |
| DEFAULT | Assigns default value | `Status VARCHAR(20) DEFAULT 'Active'` |

## 7. SQL Operators

| Category | Description |
|---|---|
| Arithmetic | Mathematical calculations on numeric values |
| Comparison | Compare two values → TRUE/FALSE |
| Logical | Combine/modify multiple conditions |
| Special | Filtering, pattern matching, existence checks |
| Set | Combine results from multiple SELECT queries |

### Arithmetic Operators
| Operator | Description | Example |
|---|---|---|
| + | Adds two values | `SELECT 10 + 5;` |
| - | Subtracts values | `SELECT 10 - 5;` |
| * | Multiplies values | `SELECT 10 * 5;` |
| / | Divides values | `SELECT 10 / 5;` |
| % | Remainder after division | `SELECT 10 % 3;` |

### Comparison Operators
| Operator | Description | Example |
|---|---|---|
| = | Equal to | `WHERE Salary = 50000` |
| != | Not equal to | `WHERE Salary != 50000` |
| <> | Not equal to (standard SQL) | `WHERE Salary <> 50000` |
| > | Greater than | `WHERE Salary > 50000` |
| < | Less than | `WHERE Salary < 50000` |
| >= | Greater than or equal to | `WHERE Salary >= 50000` |
| <= | Less than or equal to | `WHERE Salary <= 50000` |

### Logical Operators
| Operator | Description | Example |
|---|---|---|
| AND | TRUE if all conditions true | `WHERE Dept='IT' AND Salary>50000` |
| OR | TRUE if at least one condition true | `WHERE Dept='IT' OR Dept='HR'` |
| NOT | Reverses a condition's result | `WHERE NOT Dept='IT'` |

### Special Operators
| Operator | Description | Example |
|---|---|---|
| IN | Value exists in a list | `WHERE Dept IN ('IT','HR')` |
| BETWEEN | Value falls within a range | `WHERE Salary BETWEEN 30000 AND 60000` |
| LIKE | Pattern search in text | `WHERE Name LIKE 'A%'` |
| IS NULL | Checks for NULL values | `WHERE Phone IS NULL` |
| EXISTS | Checks if subquery returns rows | `WHERE EXISTS (SELECT 1 FROM Orders)` |

### Set Operators
| Operator | Description | Example |
|---|---|---|
| UNION | Combines results, removes duplicates | `SELECT City FROM A UNION SELECT City FROM B` |
| UNION ALL | Combines results, keeps duplicates | `SELECT City FROM A UNION ALL SELECT City FROM B` |
| INTERSECT | Common rows from both queries | `SELECT City FROM A INTERSECT SELECT City FROM B` |
| EXCEPT | Rows in first query not in second | `SELECT City FROM A EXCEPT SELECT City FROM B` |

## 8. Joins in SQL
Combine data from two or more tables based on a related column (usually PK & FK).

| Join Type | Description | Example |
|---|---|---|
| INNER JOIN | Only matching records from both tables | `SELECT * FROM A INNER JOIN B ON A.id = B.id;` |
| LEFT JOIN | All records from left + matches from right | `SELECT * FROM A LEFT JOIN B ON A.id = B.id;` |
| RIGHT JOIN | All records from right + matches from left | `SELECT * FROM A RIGHT JOIN B ON A.id = B.id;` |
| FULL JOIN | All records when there's a match in either table | `SELECT * FROM A FULL JOIN B ON A.id = B.id;` |
| CROSS JOIN | Every possible combination of rows | `SELECT * FROM A CROSS JOIN B;` |
| SELF JOIN | Table joined with itself | `SELECT * FROM Emp E1 JOIN Emp E2 ON E1.ManagerID = E2.EmpID;` |

## 9. CASE Statement
Adds conditional logic (like IF-ELSE / switch-case).

```sql
CASE
  WHEN condition THEN result
  ELSE default-result
END
```

| Keyword | Purpose |
|---|---|
| CASE | Starts conditional logic |
| WHEN | Specifies condition/value |
| THEN | Result if condition is true |
| ELSE | Default result if no match |
| END | Ends the CASE statement |

**Types:**

| Type | Description | Example |
|---|---|---|
| Simple CASE | Compares one expression to multiple values | `CASE DeptID WHEN 101 THEN 'IT' WHEN 102 THEN 'HR' END` |
| Searched CASE | Evaluates multiple conditions via WHEN | `CASE WHEN Salary > 50000 THEN 'High' ELSE 'Low' END` |

## 10. Window Functions
Perform calculations across a set of related rows **while keeping individual rows** (unlike GROUP BY, which collapses rows).

### Ranking Functions
| Function | Description | Example |
|---|---|---|
| ROW_NUMBER() | Unique sequential number per row | `ROW_NUMBER() OVER(ORDER BY Salary DESC)` |
| RANK() | Rank with gaps for ties | `RANK() OVER(ORDER BY Salary DESC)` |
| DENSE_RANK() | Rank without gaps for ties | `DENSE_RANK() OVER(ORDER BY Salary DESC)` |
| NTILE(n) | Divides rows into n equal groups | `NTILE(4) OVER(ORDER BY Salary)` |

### Analytical Functions
| Function | Description | Example |
|---|---|---|
| LAG() | Value from previous row | `LAG(Salary) OVER(ORDER BY EmpID)` |
| LEAD() | Value from next row | `LEAD(Salary) OVER(ORDER BY EmpID)` |

### Aggregate Window Functions
| Function | Description | Example |
|---|---|---|
| SUM() OVER() | Running/cumulative total | `SUM(Salary) OVER(ORDER BY EmpID)` |
| AVG() OVER() | Running average | `AVG(Salary) OVER(ORDER BY EmpID)` |
| COUNT() OVER() | Counts rows in window | `COUNT(*) OVER(PARTITION BY DeptID)` |
| MIN() OVER() | Minimum in window | `MIN(Salary) OVER(PARTITION BY DeptID)` |
| MAX() OVER() | Maximum in window | `MAX(Salary) OVER(PARTITION BY DeptID)` |

### Value Functions
| Function | Description | Example |
|---|---|---|
| FIRST_VALUE() | First value in window | `FIRST_VALUE(Salary) OVER(ORDER BY Salary)` |
| LAST_VALUE() | Last value in window | `LAST_VALUE(Salary) OVER(ORDER BY Salary)` |
| NTH_VALUE() | Nth value in window | `NTH_VALUE(Salary,2) OVER(ORDER BY Salary)` |
