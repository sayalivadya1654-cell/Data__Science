## 1. Creating 1D array
- An array is a collection of elements of the same data type
- A 1D array contains data in a single row (one list).
- It has only one axis.
- Syntax:

```python
# syntax
import numpy as np
arr = np.array([10, 20, 30, 40])
```

## 2. Creating 2D Array
- A 2D array is a list of lists.
- It represents data in rows and columns, like an Excel sheet.
- Syntax:

```python
import numpy as np
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
```

```python
# 2D array :
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
arr2
```

## 3. Creating 3d array
- Multiple 2D arrays stacked together

```python
arr = np.array([
    [               # Matrix 0
        [1, 2],     # Row 0
        [3, 4]      # Row 1
    ],

    [               # Matrix 1
        [5, 6],     # Row 0
        [7, 8]      # Row 1
    ]
])
arr
```

# NumPy Array Indexing & Slicing

## 1. Indexing
- Indexing means accessing one or more specific elements from an array using their position (index).
- Every element in a NumPy array has a unique position called an index.
- Index numbering always starts from 0.
- Indexing is used to retrieve a single element or modify its value.

## Types of Indexing
1. Positive Indexing
2. Negative Indexing
3. 1D Indexing
4. 2D Indexing
5. 3D Indexing
6. Fancy Indexing
7. Boolean Indexing


```python
# 1. positive indexing in 1D array
# Starts from left to right.
arrr = np.array([10,20,30,40,50])
print(arrr)
print(arrr[0])
print(arrr[2])
```

```python
# 2. Negative Indexing in 2D array
# Starts from right to left.
print(arr1[-1])
print(arr1[-3])
print(arr1[2:-3])
print(arr1[::-1])
```

```python
# example of both
print(arr1)
print(arr1[0])
print(arr1[2:4])
print(arr1[-5:])
print(arr1[::-1])    # negative indexing only
print(arr1[1:4])
```

```python
# 3. 2D Array Indexing
# A 2D array contains rows and columns.
ar = np.array([
    [10,20,30],
    [30,40,50],
    [50,80,10]
])
print(ar[0,0])
print(ar[1,1])
print(ar[2,2])
```
