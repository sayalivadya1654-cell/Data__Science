## 1. Creating 1D array
- An array is a collection of elements of the same data type
- A 1D array contains data in a single row (one list).
- It has only one axis.
- Syntax:

```python
# syntax
import numpy as np
arr = np.array([10, 20, 30, 40])
```

```python
# creating array
# 1D array
arr1 = np.array([1,2,3,4,5])
arr1
```

```python
arr = np.array(["sayali", 22, 2026])
arr
```

## 2. Creating 2D Array
- A 2D array is a list of lists.
- It represents data in rows and columns, like an Excel sheet.
- Syntax:

```python
import numpy as np
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
```

```python
# 2D array :
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
arr2
```

## 3. Shape
- shape tells the size of each dimension.
- It returns the number of rows and columns.
- (output--> shape: (row, column))
- Syntax:


```python
# syntax
arr.shape
```

```python
# array attributes:
print("shape:", arr2.shape)  # output--> shape: (row, column)
```

## dtype (Data Type)
- dtype tells the type of data stored in the array.
- Common Data Types
1. int32 / int64	    ---> Integer numbers
2. float32 / float64	---> Decimal numbers
3. bool	              ---> True or False
4. str / Unicode      ---> (<U)	Text values
- Syntax:

```python
# syntax
arr.dtype
```

```python
# data types:
arr2.dtype
```

## size
- size returns the total number of elements in the array.
- Syntax:

```python
# syntax
arr.size
```

```python
# size:
arr2.size
```

# simple array operations

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
print("addition :",a + b)
print("multiplication : ",a * b)
print("division :",a / b)
print("substraction :",a - b)
```
